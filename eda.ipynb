### Are there outliers

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
for column in df.columns:
    if df[column].dtype in ['int64', 'float64']:
        plt.figure()
        sns.boxplot(x=df[column])
        plt.title(f'Boxplot of {column}')
        plt.show()
```
